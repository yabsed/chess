# 1. What is dataclass?

```python
class PlainPlayer: 
    def __init__(self, name: str, rating: int = 1200):
        self.name   = name 
        self.rating = rating 

plain = PlainPlayer("Jimin")
print(plain.name, plain.rating)
```

```python
from dataclasses import dataclass 

@dataclass 
class Player: 
    name   : str 
    rating : int = 1200 

player = Player("Jimin")         # __init__
print(player)                    # __repr__
print(player == Player("Jimin")) # __eq__
```

```python
beginner    = Player("Jimin")
experienced = Player("Minsu", rating=1500)

beginner.rating += 20 

print(beginner   .rating)
print(experienced.rating)
```

```python
from dataclasses import field 

@dataclass 
class GameRecord: 

    moves    : list[str] = field(default_factory = list)
    revision : int = 0

    def add_move(self, move: str) -> None: 
        self.moves.append(move) 
        self.revision += 1 

first, second = GameRecord(), GameRecord() 
first.add_move("e4")

print(first)
print(second)
```

# 2. What is Literal?

```python
from typing import Literal 

Color = Literal["black", "white"]
```

```python
color_str: str = "white"
color_str      = "nonsense" # editor does not care 
print(color_str)

color_lit: Color = "white"
color_lit        = "black"
color_lit        = "nonsense" # type-chekced 

print(color_lit) # but no runtime-check
```

```python
def set_turn(color: Color) -> str:
    return f"{color} 차례입니다." 

print(set_turn("white")) # editor let me know only white/black is feasible
print(set_turn("nonsense")) # type-checked, but no runtime-check

def set_turn_str(color: str) -> str:
    return f"{color} 차례입니다."

print(set_turn_str("white")) # i have to remember the essense of value color 
print(set_turn_str("nonsense")) # editor does not care
```

```python
from typing import get_args 

print(get_args(Color))

def validate_color(color: str) -> bool:
    return color in get_args(Color)

print(validate_color("nonsense"))
```
